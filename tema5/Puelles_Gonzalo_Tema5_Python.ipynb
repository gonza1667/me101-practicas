```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("estudiantes_limpio.csv")
df.head()
```

```python
media = df["nota"].mean()
mediana = df["nota"].median()
std_muestral = df["nota"].std()                  # pandas usa ddof=1
var_poblacional = np.var(df["nota"].dropna())    # numpy usa ddof=0
q1 = df["nota"].quantile(0.25)
q3 = df["nota"].quantile(0.75)
iqr = q3 - q1

print(f"Media:            {media:.4f}")
print(f"Mediana:          {mediana:.4f}")
print(f"Desv. std (mues): {std_muestral:.4f}")
print(f"Var. poblacional: {var_poblacional:.4f}")
print(f"Q1:               {q1:.4f}")
print(f"Q3:               {q3:.4f}")
print(f"IQR:              {iqr:.4f}")
```

```python
skew_pd = df["nota"].skew()
kurt_pd = df["nota"].kurt()
skew_sp = stats.skew(df["nota"].dropna())
kurt_sp = stats.kurtosis(df["nota"].dropna())

print(f"Skewness pandas: {skew_pd:.4f}")
print(f"Kurtosis pandas: {kurt_pd:.4f}")
print(f"Skewness scipy:  {skew_sp:.4f}")
print(f"Kurtosis scipy:  {kurt_sp:.4f}")
```

```python
def resumen_estadistico(serie, decimales=4):
    serie = serie.dropna()
    n = int(serie.count())
    media = serie.mean()
    mediana = serie.median()
    desv_std = serie.std()
    cv_pct = desv_std / media * 100
    return {
        "n": n,
        "media": round(media, decimales),
        "mediana": round(mediana, decimales),
        "desv_std": round(desv_std, decimales),
        "cv_pct": round(cv_pct, decimales),
    }

print(resumen_estadistico(df["nota"]))
```

```python
def clasificar_dispersion(cv_pct):
    if cv_pct < 15:
        return "Baja"
    elif cv_pct < 30:
        return "Moderada"
    else:
        return "Alta"

for columna in ["nota", "asistencia_pct"]:
    res = resumen_estadistico(df[columna])
    nivel = clasificar_dispersion(res["cv_pct"])
    print(f"--- {columna} ---")
    print(f"  n:           {res['n']}")
    print(f"  Media:       {res['media']}")
    print(f"  Mediana:     {res['mediana']}")
    print(f"  Desv. std:   {res['desv_std']}")
    print(f"  CV (%):      {res['cv_pct']}")
    print(f"  Dispersión:  {nivel}")
    print()
```
