```python
import pandas as pd
import numpy as np

datos = {
    "nombre": ["Ana", "Luis", "María", "Carlos", "Sofía", "Diego", "Valeria"],
    "curso": ["ME101", "ME101", "ME102", "ME102", "ME101", "ME102", "ME101"],
    "nota": [15.5, 12.0, np.nan, 17.0, 14.5, 11.0, 16.0],          # NaN en María
    "asistencia_pct": [95, 80, 88, np.nan, 92, 70, 85],            # NaN en Carlos
}

df_original = pd.DataFrame(datos)
df_original.to_csv("estudiantes.csv", index=False)
```

```python
df = pd.read_csv("estudiantes.csv")
print(df.shape)         # (filas, columnas)
print(df.isna().sum())  # NaN por columna
```

```python
# (a) eliminamos las filas con nota faltante
df_limpio = df.dropna(subset=["nota"]).copy()

# (b) rellenamos la asistencia_pct con el promedio de la columna
df_limpio["asistencia_pct"] = df_limpio["asistencia_pct"].fillna(
    df_limpio["asistencia_pct"].mean()
)

# (c) verificar que no queden faltantes
print(df_limpio.isna().sum())
print(df_limpio.shape)
```

```python
promedio_por_curso = df_limpio.groupby("curso")["nota"].mean()
print(promedio_por_curso)
```

```python
df_limpio.to_csv("estudiantes_limpio.csv", index=False)
```
