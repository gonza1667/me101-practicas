```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Reconstruye tu dataset del Tema 9
datos = {
    "horas_estudio": [2, 4, 5, 3, 6, 8, 7, 1, 9, 5],
    "nota_examen":   [9, 12, 13, 10, 15, 17, 16, 7, 18, 14],
    "horas_sueno":   [8, 7, 6, 7, 6, 5, 6, 8, 5, 7],
    "horas_celular": [6, 5, 3, 5, 3, 2, 4, 7, 2, 4],
}

df = pd.DataFrame(datos)
corr = df.corr()
print(corr.round(4))
```

```python
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm",
            center=0, vmin=-1, vmax=1, square=True,
            cbar_kws={"label": "Correlación de Pearson"})
plt.title("Matriz de correlación (seaborn)")
plt.tight_layout()
plt.savefig("heatmap_seaborn.png", dpi=150)
plt.show()
```

```python
mascara = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=(6, 5))
sns.heatmap(corr, mask=mascara, annot=True, fmt=".2f", cmap="coolwarm",
            center=0, vmin=-1, vmax=1, square=True,
            cbar_kws={"label": "Correlación de Pearson"})
plt.title("Matriz de correlación con máscara (seaborn)")
plt.tight_layout()
plt.savefig("heatmap_seaborn_mascara.png", dpi=150)
plt.show()

print("Celdas visibles:", int((~mascara).sum()))
```
