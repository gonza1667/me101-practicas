```python
import numpy as np

notas = np.array([11, 14, 20, 8, 16, 13])

print("shape:", notas.shape)
print("dtype:", notas.dtype)
print("ndim:", notas.ndim)
```

```python
# Completa: promedio, maximo, y cantidad de aprobados
promedio = notas.mean()
maximo = notas.max()
cantidad_aprobados = (notas >= 10.5).sum() # pista: suma un array booleano

print("Promedio:", promedio)
print("Maximo:", maximo)
print("Aprobados:", cantidad_aprobados)
```

```python
import pandas as pd

datos = {
    "nombre": ["Gonzalo", "Damaris", "Diego", "Marcela", "Rodrigo", "Zahir"],
    "nota": [18, 14, 20, 8, 16, 13],
    "asistencia_pct": [85, 60, 92, 75, 68, 90],
}
df = pd.DataFrame(datos)

filtrados = df[(df["nota"] >= 10.5) & (df["asistencia_pct"] >= 70)]
print(filtrados)
```
