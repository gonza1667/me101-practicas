```python
notas = [14, 16, 12, 18, 9]          # list
print(notas[0], notas[-1])           # 14 9 -- primer y último elemento
notas.append(20)                     # las listas son mutables
print(notas)                         # [14, 16, 12, 18, 9, 20]

punto = (13.52, -71.97)              # tuple: inmutable
# punto[0] = 0  -- esto lanzaría TypeError, descomentar para demostrarlo

estudiante = {"nombre": "Ana", "nota": 16, "ciclo": 3}   # dict
print(estudiante["nota"])            # 16 -- acceso por clave
```

```python
import numpy as np

notas_np = np.array([14, 16, 12, 18, 9])
print(notas_np.shape, notas_np.dtype, notas_np.ndim)
```

```python
notas = [14, 16, 12, 18, 9]   # restauramos la lista original (5 elementos)

# Con lista: requiere bucle o list comprehension
notas_lista_mas_1 = [n + 1 for n in notas]

# Con NumPy: vectorizado, sin bucle explícito
notas_np_mas_1 = notas_np + 1

print(notas_lista_mas_1)
print(notas_np_mas_1)
```

```python
evaluaciones = np.array([[14, 16], [12, 10], [18, 19]])
print(evaluaciones.shape)          # (3, 2): 3 filas, 2 columnas
print(evaluaciones[1, 0])          # 12 -- fila 1 (segundo estudiante), columna 0
print(evaluaciones.mean(axis=1))   # promedio por fila (por estudiante)
```

```python
import pandas as pd

datos = {
    "nombre": ["Ana", "Luis", "Marco", "Rosa", "Iván"],
    "nota": [16, 12, 18, 9, 14],
    "ciclo": [3, 5, 3, 1, 2],
}
df = pd.DataFrame(datos)

print(df.shape)
print(df.dtypes)
df.head()
```

```python
print(df['nota'].mean())      # 13.8
print(df.loc[0])              # primera fila, por etiqueta
print(df.iloc[0])             # primera fila, por posición (mismo resultado aquí)

aprobados = df.loc[df['nota'] >= 10.5]
print(aprobados[['nombre', 'nota']])
```
